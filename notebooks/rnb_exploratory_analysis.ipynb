Project RnB Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
df = pd.read_csv('RnB_merged_clean.csv')
```

```python
df.info()
df.shape
df.head()
df.describe()
```

Univariate Analysen - Verteilungen einzelner Variablen:

```python
# Beispiel: Histogramm für eine numerische Variable

# Daten filtern
tempo_90s = df[df["era"] == "90s"]["tempo"]
tempo_10s = df[df["era"] == "10s"]["tempo"]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 90s Histogramm
axes[0].hist(tempo_90s, bins=30, color="skyblue", edgecolor="black")
axes[0].set_title("Tempo – 90s")
axes[0].set_xlabel("Tempo")
axes[0].set_ylabel("Anzahl der Songs")

# 10s Histogramm
axes[1].hist(tempo_10s, bins=30, color="salmon", edgecolor="black")
axes[1].set_title("Tempo – 10s")
axes[1].set_xlabel("Tempo")
axes[1].set_ylabel("TAnzahl der Songs")

plt.tight_layout()
plt.show()
```

```python
#Boxplot für eine numerische Variable

df_90s = df[df["era"] == "90s"]
df_10s = df[df["era"] == "10s"]

plt.figure(figsize=(14, 6))

# --- 90s ---
plt.subplot(1, 2, 1)
sns.boxplot(x=df_90s['loudness'], color="skyblue")
plt.title('Loudness – 90s')
plt.xlabel('loudness')
plt.xlim(-25, 0)  

# --- 10s ---
plt.subplot(1, 2, 2)
sns.boxplot(x=df_10s['loudness'], color="salmon")
plt.title('Loudness – 10s')
plt.xlabel('loudness')
plt.xlim(-25, 0)

plt.tight_layout()
plt.show()
```
